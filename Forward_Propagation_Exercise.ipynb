1) Define a ReLU activation functions that accepts a vector x as an input and return an activation value. Then, print the output value of the ReLU function for x=[0.5 -1.4]'

np.maximum() is a NumPy function that compares two arrays (or an array and a scalar) element-wise and returns the larger value at each position.This means for any **positive input, it returns the input** itself, and for any **negative input, it returns 0**.
image.png

```python
def relu(x):
 return np.maximum(0, x)
```

```python
#Print the output of a ReLU function
x = np.array([0.5, -1.4])
y = relu(x)
print(f"The input of a ReLU function is: {x}")
print(f"The output of a ReLU function is: {y}")
```

2) Define a sigmoid activation functions that accepts a vector x as an input and return an activation value.  Then, print the output value of the sigmoid function for x=[0.5 -1.4]'

image.png

```python
def sigmoid(x):
 return 1 / (1 + np.exp(-x))
```

```python
#Print the output of a sigmoid function
y = sigmoid(x)
print(f"The output of a sigmoid function is: {y}")
```

3) Define a forward propagation function that:

    i) Accepts input data vector, weights and biases
    ii) Return the outputs at each layer
    iii) Use the function to calculate the output of the hidden neurons and then the output neuron for the given NN architecture.

```python
def forward_prop(X, W, b):
    Z = np.dot(W, X) + b
    return Z

#Hidden layer calculation
W1 = np.array ([[0.1, -0.3],
               [0.2,  0.4],
               [-0.5, 0.6]]) # W1 is a (3 x 2) matrix - 3 hidden neurons and 2 input features
print(f"W1 = {W1}, shape of W1 is {np.shape(W1)}")

b1 = np.array ([0.1]) # b1 is the bias value going to all three hidden neurons
print(f"b1 = {b1}, shape of b1 is {np.shape(b1)}")

Z1 = forward_prop(X.T, W1, b1)
A1 = relu(Z1)  # Activation function
print(f"The output of a hidden layer, A1, is: {A1}")


#Output Layer calculation
W2 = np.array ([[0.5, 3.3, 5.5]]) # W2 is a (1 x 3) matrix - 1 output neuron from 3 hidden neurons
print(f"W2 = {W2}, shape of W2 is {np.shape(W2)}")

b2 = np.array ([0.1]) # b2 is the bias value going to the output neuron
print(f"b2 = {b2}, shape of b2 is {np.shape(b2)}")

Z2 = forward_prop(A1, W2, b2)
A2 = sigmoid(Z2)
print(f"The output of a hidden layer after ReLU is: {A2}")
```
